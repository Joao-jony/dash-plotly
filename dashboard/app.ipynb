```python
import pandas as pd
import numpy as np


df_vendas = pd.read_csv("../csv_arq/sales.csv")
df_vendas["Data da Encomenda"] = pd.to_datetime(df_vendas["Data da Encomenda"], format="%d/%m/%Y")
df_vendas["Data de Expedição"] = pd.to_datetime(df_vendas["Data de Expedição"], format="%d/%m/%Y")
display(df_vendas.info())
```

```python
df_receitas_total = df_vendas.groupby("Região", as_index=False)["Receita Total"].sum().sort_values("Receita Total", ascending=False)
display(df_receitas_total)
```

```python
df_receita_total_itens = df_vendas.groupby("Tipo de Item", as_index=False)[["Receita Total", "Lucro Total"]].sum().sort_values("Receita Total", ascending=False)
display(df_receita_total_itens)
```

```python
df_receita_canal = df_vendas.groupby("Canal de Venda", as_index=False)["Receita Total"].sum()
display(df_receita_canal)
```

```python
df_receita_por_prioridade = df_vendas.groupby("Prioridade da Encomenda", as_index=False)["Receita Total"].sum()
display(df_receita_por_prioridade)
```

```python

df_data = df_vendas

df_data["Mes"] = df_data["Data da Encomenda"].dt.to_period("M").astype(str)
df_data = df_data.groupby("Mes", as_index=False)[["Receita Total", "Lucro Total", "Custo Total"]].sum().sort_values("Mes")


display(df_data)
```

```python
df_recieta_pais = df_vendas.groupby("País", as_index=False)["Receita Total"].sum().sort_values("Receita Total", ascending=False).head(15)
display(df_recieta_pais)
```

```python
df_receita_regiao_pais_item = df_vendas.groupby(["Região", "Tipo de Item"], as_index=False)["Receita Total"].sum()
display(df_receita_regiao_pais_item)
```

```python
df_receita_preco_custo_margem = df_vendas.drop(columns=["Região", "Canal de Venda", "Prioridade da Encomenda", "Mes", "ID da Encomenda", "Data da Encomenda", "Data de Expedição", "Lucro Total", "Custo Total"])
display(df_receita_preco_custo_margem)
```

```python
df_receita_lucro_custo_total = df_vendas.groupby("Região", as_index=False)[["Receita Total", "Lucro Total", "Custo Total"]].sum().sort_values("Receita Total", ascending=False)
display(df_receita_lucro_custo_total)

df_rcl = df_receita_lucro_custo_total.melt(id_vars='Região', var_name='Categoria', value_name='Valor')
df_rcl = df_rcl.groupby(["Região", "Categoria"], as_index=False)["Valor"].sum()
display(df_rcl)
```

```python
df_por_pais = df_vendas.groupby("País", as_index=False)[["Receita Total", "Lucro Total", "Custo Total"]].sum().sort_values("Receita Total", ascending=False).head(10)
df_rcl_por_pais = df_por_pais.melt(id_vars='País', var_name='Categoria', value_name='Valor')
df_rcl_por_pais = df_rcl_por_pais.groupby(["País", "Categoria"], as_index=False)["Valor"].sum()

display(df_rcl_por_pais)

display(df_vendas["País"].unique())
```

```python

df_vendas["Data da Encomenda"] = pd.to_datetime(df_vendas["Data da Encomenda"], format="%d/%m/%Y", errors="coerce")
df_vendas["Data de Expedição"] = pd.to_datetime(df_vendas["Data de Expedição"], format="%d/%m/%Y", errors="coerce")

df_vendas["Dias"] = ( df_vendas["Data de Expedição"] - df_vendas["Data da Encomenda"]).dt.days

display(df_vendas[["Dias"]])


```

```python
df_vendas["Margens"] = (df_vendas["Lucro Total"] / df_vendas["Receita Total"] * 100).round(2)
display(df_vendas["Margens"])

df_margem = df_vendas.groupby("Região", as_index=False)["Margens"].mean().sort_values("Margens", ascending=False)

display(df_margem)
```

```python
df_receita_pais = df_vendas.groupby("País", as_index=False)["Receita Total"].sum().sort_values("Receita Total", ascending=False).head(15)
display(df_receita_pais)
```
